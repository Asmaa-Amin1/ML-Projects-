```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

```python
california_data = fetch_california_housing()
```

```python
df = pd.DataFrame(california_data['data'],
                  columns=california_data['feature_names'])
```

```python
df['MedHouseVal'] = california_data['target']
```

```python
X = df.drop(['AveBedrms','MedHouseVal'] , axis=1)
y=df['MedHouseVal']
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
 
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2005)
```

```python
x_train
```

```python
from sklearn.linear_model import LinearRegression

model_lr = LinearRegression()
model_lr.fit(x_train,y_train) # return best a,b
```

```python
model_lr.coef_
```

```python
model_lr.intercept_
```

```python
y_pred_lr = model_lr.predict(x_test) # y = aX + b
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error

mse_lr = mean_squared_error(y_test,y_pred_lr)
mae_lr = mean_absolute_error(y_test,y_pred_lr)

print(f'MSE: {mse_lr}')
print(f'MAE: {mae_lr}')
```

```python
from sklearn.linear_model import SGDRegressor
model_sgd = SGDRegressor(
    max_iter=100000, eta0=1e-9,
    learning_rate='adaptive' #invscaling #adaptive
    ,random_state=2005,
    power_t=1e-3
)

model_sgd.fit(x_train,y_train)
```

```python
y_pred_sgd = model_sgd.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error

mse_sgd = mean_squared_error(y_test,y_pred_sgd)
mae_sgd = mean_absolute_error(y_test,y_pred_sgd)

print(f'MSE: {mse_sgd}')
print(f'MAE: {mae_sgd}')
```

```python
# Polynomial Regrission
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2 , include_bias=False)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.fit_transform(x_test)
```

```python
x_train.shape
```

```python
x_train_poly.shape
```

```python
model_lr_poly = LinearRegression()
model_lr_poly.fit(x_train_poly,y_train)
```

```python
y_pred_poly=model_lr_poly.predict(x_test_poly)
```

```python
model_lr_poly.coef_
```

```python
model_lr_poly.intercept_
```

```python
mse_poly = mean_squared_error(y_test,y_pred_poly)
mae_poly = mean_absolute_error(y_test,y_pred_poly)

print(f'MSE: {mse_poly}')
print(f'MAE: {mae_poly}')
```

```python
from sklearn.linear_model import Ridge

model_ridge_lr = Ridge(alpha=1000000000)
model_ridge_lr.fit(x_train,y_train)
y_pred_ridge_lr = model_ridge_lr.predict(x_test)

mse_ridge_lr = mean_squared_error(y_test,y_pred_ridge_lr)
mae_ridge_lr = mean_absolute_error(y_test,y_pred_ridge_lr)

print(f'MSE: {mse_ridge_lr}')
print(f'MAE: {mae_ridge_lr}')
```

```python
from sklearn.linear_model import Ridge

model_ridge_poly = Ridge(alpha=1000000000)
model_ridge_poly.fit(x_train_poly,y_train)
y_pred_ridge_poly = model_ridge_poly.predict(x_test_poly)

mse_ridge_poly = mean_squared_error(y_test,y_pred_ridge_poly)
mae_ridge_poly = mean_absolute_error(y_test,y_pred_ridge_poly)

print(f'MSE: {mse_ridge_poly}')
print(f'MAE: {mae_ridge_poly}')
```

```python
from sklearn.linear_model import Lasso

model_lasso_lr = Ridge(alpha=1000)
model_lasso_lr.fit(x_train,y_train)
y_pred_lasso_lr = model_lasso_lr.predict(x_test)

mse_lasso_lr = mean_squared_error(y_test,y_pred_lasso_lr)
mae_lasso_lr = mean_absolute_error(y_test,y_pred_lasso_lr)

print(f'MSE: {mse_lasso_lr}')
print(f'MAE: {mae_lasso_lr}')
```

```python
from sklearn.linear_model import Lasso

model_lasso_poly = Ridge(alpha=0.01)
model_lasso_poly.fit(x_train_poly,y_train)
y_pred_lasso_poly  = model_lasso_poly.predict(x_test_poly)

mse_lasso_poly = mean_squared_error(y_test,y_pred_lasso_lr)
mae_lasso_poly = mean_absolute_error(y_test,y_pred_lasso_lr)

print(f'MSE: {mse_lasso_poly}')
print(f'MAE: {mae_lasso_poly}')
```

```python
results = pd.DataFrame(
    {
        'Model': [
            'Linear Regression',
            'SGD',
            'Polynomial',
            'Ridge Linear',
            'Ridge Poly',
            'Lasso Linear',
            'Lasso Poly',
        ],
        'MSE': [
            mse_lr,
            mse_sgd,
            mse_poly,
            mse_ridge_lr,
            mse_ridge_poly,
            mse_lasso_lr,
            mse_lasso_poly,
        ],
        'MAE': [
            mae_lr,
            mae_sgd,
            mae_poly,
            mae_ridge_lr,
            mae_ridge_poly,
            mae_lasso_lr,
            mae_lasso_poly,
        ],
    }
)
```
